We will use the **Ridge Regression** model. Ridge regression is a form of linear regression that includes regularization (L2), which is essential to prevent **overfitting** when dealing with a large number of features, particularly the hundreds of binary variables generated from our olfactory notes (Top, Middle, Base Notes).

#### Key Metric
The **`weighted_rating`** remains our core target variable (Y). 
This metric corrects for popularity bias, ensuring our model focuses purely on predicting **sustained quality excellence**.
